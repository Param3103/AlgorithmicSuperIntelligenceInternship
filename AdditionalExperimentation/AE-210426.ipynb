```python
from datasets import load_dataset

dataset = load_dataset("openai/gsm8k", "main")
train_set = dataset["train"]
for i in range(10):
    print(train_set["question"][i])
    print()
    print(train_set["answer"][i])
    print("--------------------------")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

def get_similarity(a, b):
    emb = model.encode([a, b])
    return cosine_similarity([emb[0]], [emb[1]])[0][0]
similarities = []
l = 100
for i in range(l):
    for j in range(i+1, l):
        s = get_similarity(train_set["question"][i], train_set["question"][j])
        if j < 6:
            print(train_set["question"][i])
            print(train_set["question"][j])
            print("SIMILARITY:", s)
            print()
        similarities.append((i, j, s))
        print(i, j)
```

```python
sorted_similarities = sorted(similarities, key=lambda x: x[2], reverse=True)
sorted_similarities[:20]
```

```python
# get top 3
visited_pairs = []
for i, j, s in sorted_similarities:
    for pair in visited_pairs:
        if i in pair:
            print(i, j, pair[(pair.index(i)+1)%2])
            break
            break
        elif j in pair:
            print(i, j, pair[(pair.index(j)+1)%2])
            break
            break
    else:
        visited_pairs.append((i, j, s))
        
```

```python
# Let us look at 3 14 and 94

examples_3shot = [(train_set["question"][3], train_set["answer"][3]),
(train_set["question"][14], train_set["answer"][14]),
(train_set["question"][94], train_set["answer"][94])]
examples_3shot
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen2.5-Coder-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="cuda"
)

def few_shot_prompt(question):
    global examples_3shot, tokenizer
    prompt = f"""Solve the following math problems.

                Q: {examples_3shot[0][0]}
                A: {examples_3shot[0][1]}

                Q: {examples_3shot[1][0]}
                A: {examples_3shot[1][1]}

                Q: {examples_3shot[2][0]}
                A: {examples_3shot[2][1]}


                Q: {question}
                A:"""
    messages = [{"role": "user", "content": prompt}]
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(text, return_tensors="pt")
    inputs = {k: v.to(model.device) for k, v in inputs.items()}  # ← fix is here
    
    output = model.generate(
            **inputs,
            max_new_tokens=1024,
            do_sample=True,
            temperature=0.3,
            top_p=0.95
        )
    
    input_length = inputs["input_ids"].shape[1]
    generated_tokens = output[0][input_length:]
    return tokenizer.decode(generated_tokens, skip_special_tokens=True)
```

```python
import re

def extract_correct_answer(text):
    # Get only the text after the last "Q:" block's answer
    parts = text.split("A:")
    last_answer = parts[-1]  # everything after the final "A:"
    match = re.search(r"#### (\d+)", last_answer.strip())
    print(match.group(1) if match else None)
    return match.group(1) if match else None
def extract_answer(text):
    # Get only the text after the last "Q:" block's answer
    # **\$195**
    parts = text.split("A:")
    last_answer = parts[-1]  # everything after the final "A:"
    match = re.search(r'#### (\d+)', last_answer)
    return match.group(1) if match else None



print("Q: ", train_set["question"][200])
print("A: ", train_set["answer"][200])
correct_count = 0
correct_answer = extract_correct_answer(train_set["answer"][200])
for i in range(10): # needs to be 100
    print("RUN #", i+1)
    text = few_shot_prompt(train_set["question"][200])
    print(text)
    answer = extract_answer(text)
    print(answer, correct_answer)
    if answer == correct_answer:
        correct_count += 1

print("Final Correct count:", correct_count)
print("Accuracy:", correct_count / 100)
```
